```python
df = pd.concat([pd.DataFrame(x) for x in tournament.matches.values()])
```

```python
df['dateTime'] = pd.to_datetime(df['date_id'].astype(str))
df['weekday'] = df['dateTime'].dt.weekday

anchor_weekday = 5
df['my_week'] = df['dateTime'].apply(lambda d: d - pd.Timedelta(days=(d.weekday() + 3) % 7))
df['my_weekday'] = df['my_week'].dt.weekday

df['home_week_number'] = df.groupby('home_team')['date_id'].rank()
df['away_week_number'] = df.groupby('home_team')['date_id'].rank()
assert (df['home_week'] == df['away_week']).all()
```
